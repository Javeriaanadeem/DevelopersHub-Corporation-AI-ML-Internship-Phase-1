```python
import requests
```

```python
HF_TOKEN = "YOUR_HF_TOKEN_HERE"
API_URL = "https://api-inference.huggingface.co/models/mistralai/Mistral-7B-Instruct-v0.1"
headers = {"Authorization": f"Bearer {HF_TOKEN}"}
```

```python
UNSAFE_KEYWORDS = ['suicide', 'kill myself', 'overdose', 'self harm', 'end my life']

def is_unsafe(text):
    return any(word in text.lower() for word in UNSAFE_KEYWORDS)
```

```python
def chat(user_input):
    if is_unsafe(user_input):
        return "Please reach out to a mental health professional or call emergency services immediately."
    
    prompt = f"[INST] You are a helpful medical assistant. Answer this health question simply and clearly, always advise seeing a doctor for serious issues: {user_input} [/INST]"
    
    try:
        response = requests.post(API_URL, headers=headers, json={
            "inputs": prompt,
            "parameters": {"max_new_tokens": 200, "temperature": 0.7}
        })
        result = response.json()
        if isinstance(result, list):
            text = result[0]['generated_text']
            return text.split('[/INST]')[-1].strip()
        return "Sorry, I could not process that. Please try again."
    except:
        return "Connection error. Please check your internet and token."
```

```python
print('=' * 50)
print('   General Health Query Chatbot')
print('=' * 50)
print("Type 'quit' to exit\n")

while True:
    user = input('You: ').strip()
    if not user:
        continue
    if user.lower() == 'quit':
        print('Goodbye! Stay healthy!')
        break
    print('Bot:', chat(user), '\n')
```
